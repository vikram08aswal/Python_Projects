```python
#importing libraries
import pandas as pd
import numpy as np
```

```python
#reading data
df = pd.read_csv('/workspaces/new_python_test/data/cafe_sales.csv')
```

```python
#inspecting first few rows
df.head()
```

```python
#inspecting last few rows
df.tail()
```

```python
#data structure
df.info()
```

```python
#duplicate record
print(f'Number of duplicate records: {sum(df.duplicated())}')
```

```python
#missing record
df.isna().sum()*100 / df.shape[0]
```

```python
#unique record
df_columns = set(df.columns)
exclude = {'Transaction ID'}
df_columns = df.columns.difference(exclude)

for column in df_columns:
    print(f'{column}: \n {df[column].unique()}', '\n\n')
```

```python
#replacing ERROR and UNKNOWN values
df = df.replace(['ERROR', 'UNKNOWN'], np.nan)
```

```python
#missing record after treating "ERROR" & "UNKNOWN"
df.isna().sum()*100 / df.shape[0]
```

```python
#data type converstion
df['Quantity'] = df['Quantity'].astype('Int64')
df[['Price Per Unit', 'Total Spent']] = df[['Price Per Unit', 'Total Spent']].astype('float')
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

```python
df.info()
```

```python
#dropping record where Transaction Date is not available
df = df.dropna(subset = 'Transaction Date')
df
```

```python
#replacing missing record in "Payment Method" & "Location" with "INA"
df[['Payment Method', 'Location']] = df[['Payment Method', 'Location']].fillna('INA')
```

```python
#replacing missing record in "Item" with mode value
df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
```

```python
#replacing missing record in "Quantity" with median value
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
#inspecting item "Price Per Unit"
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
#filling missing record in "Price Per Unit" with mode value of Item category
df['Price Per Unit'] = df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
#filling missing record in "Total Spent"
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

```python
#saving cleaned file
df.to_csv('/workspaces/new_python_test/data/cafe_sales (cleaned).csv', index = False)
```
